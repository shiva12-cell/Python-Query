### Key Concepts & Reference Notes
- **Dictionary Lookups:** Python `dict` provides average O(1) key lookups.
- **Safe Membership Check:** `query in phone_book` prevents `KeyError` exceptions.
- **Reading until EOF:** In production / HackerRank, use `sys.stdin` to read all query lines until stream terminates.

---

### Solution & Demonstration


```python
def phone_book_lookup(entries, queries):
    # Build phone book dictionary
    phone_book = {name: phone for name, phone in entries}
    
    # Query lookups
    for name in queries:
        if name in phone_book:
            print(f'{name}={phone_book[name]}')
        else:
            print('Not found')

# Demo
sample_entries = [('sam', '99912222'), ('tom', '11122222'), ('harry', '12299933')]
sample_queries = ['sam', 'edward', 'harry']
phone_book_lookup(sample_entries, sample_queries)
```
