### Key Concepts & Reference Notes
- **String Immutability:** Strings cannot be modified in-place; indexing assignment `s[i] = 'c'` raises a TypeError.
- **Slicing Approach:** `string[:position] + character + string[position + 1:]` creates a modified copy.
- **List Approach:** Alternatively, convert with `list(s)`, mutate the index, and rejoin with `''.join()`.

---

### Solution & Demonstration


```python
def mutate_string(string, position, character):
    # Slice string around the target position and insert new character
    return string[:position] + character + string[position + 1:]

# Demo
original = 'abracadabra'
modified = mutate_string(original, 5, 'k')
print('Original:', original)
print('Modified:', modified)
```
