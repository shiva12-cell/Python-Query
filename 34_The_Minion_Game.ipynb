# The Minion Game

**Difficulty:** Medium | **Topic:** str

- [HackerRank Challenge](https://www.hackerrank.com/challenges/the-minion-game/problem)

---

### Key Concepts & Reference Notes
- **O(n) Mathematical Shortcut:** Instead of generating all substrings (which takes $O(n^2)$ and times out), notice that a character at index `i` starts exactly `len(s) - i` substrings.
- **Vowels vs Consonants:** Kevin gets points if starting char is in `AEIOU`; otherwise Stuart gets points.
- **Comparison:** Compare final scores to print winner or `'Draw'`.

---

### Solution & Demonstration


```python
def minion_game(string):
    vowels = set('AEIOU')
    stuart_score = 0
    kevin_score = 0
    n = len(string)
    
    # Single pass O(n) calculation
    for i in range(n):
        if string[i] in vowels:
            kevin_score += n - i
        else:
            stuart_score += n - i
    
    # Print winner and score
    if kevin_score > stuart_score:
        print(f'Kevin {kevin_score}')
    elif stuart_score > kevin_score:
        print(f'Stuart {stuart_score}')
    else:
        print('Draw')

# Demo with classic sample 'BANANA'
minion_game('BANANA')
```
