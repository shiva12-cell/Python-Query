# Capitalize

**Difficulty:** Easy | **Topic:** str, format

- [HackerRank Challenge](https://www.hackerrank.com/challenges/capitalize/problem)

---

### Key Concepts & Reference Notes
- **`str.title()` Flaw:** `str.title()` changes `'12abc'` into `'12Abc'`, failing HackerRank tests.
- **Single-Space Splitting:** `s.split(' ')` preserves consecutive spaces as empty string tokens `''`.
- **`word.capitalize()`:** Correctly capitalizes only leading letters while leaving numbers completely unaffected.

---

### Solution & Demonstration


```python
def solve(s):
    # Split on literal single spaces to preserve consecutive spaces
    words = s.split(' ')
    # Capitalize each word and rejoin with single spaces
    return ' '.join(word.capitalize() for word in words)

# Demo
print(solve('chris alan'))
print(solve('12abc 34cd'))
print(solve('hello   world')) # note triple space preserved
```
