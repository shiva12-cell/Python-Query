# Find the Runner-Up Score!

**Difficulty:** Easy | **Topic:** array, list, sort, data types

- [HackerRank Challenge](https://www.hackerrank.com/challenges/find-second-maximum-number-in-a-list/problem)

---

### Key Concepts & Reference Notes
- **Removing Duplicates with `set()`:** Multiple people can share the highest score; `set()` keeps only distinct values.
- **Sorting Descending:** `sorted(..., reverse=True)` places 1st max at index 0 and 2nd max at index 1.
- **Index `1`:** Returns the second highest unique value.

---

### Solution & Demonstration


```python
def find_runner_up(arr):
    # Convert to set for unique scores, sort descending, pick runner-up
    unique_scores = sorted(set(arr), reverse=True)
    return unique_scores[1]

# Demo
scores = [2, 3, 6, 6, 5]
print('Runner-up score:', find_runner_up(scores))
```
