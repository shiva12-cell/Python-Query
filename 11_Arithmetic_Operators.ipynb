### Key Concepts & Reference Notes
- **Standard Math Operators:** `+` (addition), `-` (subtraction), `*` (multiplication).
- **Arbitrary Precision:** Python integers never overflow; they automatically expand in memory.
- **Sequential Output:** Each `print()` statement prints on its own line.

---

### Solution & Demonstration


```python
def run_arithmetic(a, b):
    # Sum
    print(a + b)
    # Difference (a - b)
    print(a - b)
    # Product
    print(a * b)

# Demo
run_arithmetic(3, 2)
```
