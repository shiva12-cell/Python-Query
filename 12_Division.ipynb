### Key Concepts & Reference Notes
- **Integer (Floor) Division (`//`):** Divides and rounds down to the nearest whole integer.
- **Float Division (`/`):** Performs standard division and always returns a floating-point number in Python 3.
- **Precision:** Float division provides IEEE 754 double precision.

---

### Solution & Demonstration


```python
def show_division(a, b):
    # Integer floor division
    print(a // b)
    # Floating-point division
    print(a / b)

# Demo
show_division(4, 3)
```
