---

### Key Concepts & Reference Notes
- **Logical Flow:** Translating business requirements directly to conditional branches.
- **Modulo Operator `%`:** `n % 2 != 0` checks for odd numbers.
- **Inclusive Ranges:** Python allows `6 <= n <= 20` directly.

---

### Solution & Demonstration


```python
def evaluate_number(n):
    if n % 2 != 0:
        print('Weird')
    elif 2 <= n <= 5:
        print('Not Weird')
    elif 6 <= n <= 20:
        print('Weird')
    else:
        print('Not Weird')

# Demo
evaluate_number(3)   # Odd -> Weird
evaluate_number(24)  # Even > 20 -> Not Weird
```
