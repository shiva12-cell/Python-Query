# Merge the Tools!

**Difficulty:** Medium | **Topic:** str, format, data types

---

### Key Concepts & Reference Notes
- **Chunking by Slicing:** Step through string in increments of `k`: `string[i : i + k]`.
- **Order-Preserving Deduplication:** In Python 3.7+, `dict.fromkeys(chunk)` deduplicates keys while preserving original insertion order.
- **Linear Performance:** Runs in O(n) total time with no nested search overhead.

---

### Solution & Demonstration


```python
def merge_the_tools(string, k):
    # Slice string into consecutive non-overlapping chunks of length k
    for i in range(0, len(string), k):
        chunk = string[i : i + k]
        # dict.fromkeys eliminates duplicate characters while preserving order
        print(''.join(dict.fromkeys(chunk)))

# Demo
merge_the_tools('AABCAAADA', 3)
```
