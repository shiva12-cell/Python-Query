# Find a string

**Difficulty:** Easy | **Topic:** str

- [HackerRank Challenge](https://www.hackerrank.com/challenges/find-a-string/problem)

---

### Key Concepts & Reference Notes
- **Overlapping Occurrences:** Standard `str.count()` does NOT count overlaps (e.g. `CDC` in `CDCDC`).
- **Sliding Window:** Loop through all possible starting indices up to `len(string) - len(sub) + 1`.
- **Slice Comparison:** Check `string[i : i + len(sub)] == sub`.

---

### Solution & Demonstration


```python
def count_substring(string, sub_string):
    count = 0
    sub_len = len(sub_string)
    # Scan all sliding windows of length sub_len
    for i in range(len(string) - sub_len + 1):
        if string[i : i + sub_len] == sub_string:
            count += 1
    return count

# Demo with overlapping test case
text = 'ABCDCDC'
pattern = 'CDC'
print(f'Occurrences of "{pattern}" in "{text}":', count_substring(text, pattern))
```
