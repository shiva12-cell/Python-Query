### Key Concepts & Reference Notes
- **Tuples are Immutable:** Unlike lists, tuples cannot be modified once created.
- **Hashability:** Because tuples are immutable, they are hashable with `hash(t)`.
- **Conversion:** `tuple(iterable)` builds a tuple directly from any iterable.

---

### Solution & Demonstration


```python
def compute_tuple_hash(elements):
    # Convert list of integers to a tuple
    t = tuple(elements)
    # Compute hash value
    print(hash(t))

# Demo
compute_tuple_hash([1, 2])
```
