### Key Concepts & Reference Notes
- **Parity Checking:** `n % 2 != 0` checks if a number is odd.
- **Chained Comparison:** Python supports mathematical range checking like `2 <= n <= 5`.
- **Branch Evaluation:** Evaluates sequentially; the first matching branch executes.

---

### Solution & Demonstration


```python
def check_weird(n):
    # If n is odd -> Weird
    if n % 2 != 0:
        print('Weird')
    # If n is even and between 2 and 5 (inclusive) -> Not Weird
    elif 2 <= n <= 5:
        print('Not Weird')
    # If n is even and between 6 and 20 (inclusive) -> Weird
    elif 6 <= n <= 20:
        print('Weird')
    # If n is even and greater than 20 -> Not Weird
    else:
        print('Not Weird')

# Test on multiple representative numbers
for test_val in [3, 4, 18, 24]:
    print(f'n = {test_val} -> ', end='')
    check_weird(test_val)
```
