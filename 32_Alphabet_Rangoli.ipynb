# Alphabet_Rangoli

**Difficulty:** Easy | **Topic:** str, format

---

### Key Concepts & Reference Notes
- **`string.ascii_lowercase`:** Access all 26 lowercase English letters cleanly.
- **Palindromic Letter Wings:** Slicing `alphabet[size - 1 : i : -1] + alphabet[i:size]` creates decreasing then increasing letters.
- **Line Centering:** Each row is joined with `-` and centered within `4 * size - 3` characters.

---

### Solution & Demonstration


```python
import string

def print_rangoli(size):
    alpha = string.ascii_lowercase
    lines = []
    
    # Generate each line from center outward
    for i in range(size):
        row_letters = alpha[size - 1 : i : -1] + alpha[i:size]
        row_str = '-'.join(row_letters)
        lines.append(row_str.center(4 * size - 3, '-'))
    
    # Assemble full rangoli: top half (reversed) + bottom half
    full_rangoli = lines[::-1] + lines[1:]
    print('\n'.join(full_rangoli))

# Demo for size = 3
print_rangoli(3)
```
