### Key Concepts & Reference Notes
- **Finding Second Lowest:** Extract grades, use `set()` to drop duplicates, sort ascending, and choose index `1`.
- **Filtering:** List comprehension `[name for name, score in students if score == second_lowest]`.
- **Alphabetical Ordering:** `sorted(names)` ensures names are printed in lexicographical order.

---

### Solution & Demonstration


```python
def second_lowest_students(records):
    # 1. Find second lowest unique grade
    unique_grades = sorted(set(score for name, score in records))
    second_lowest_grade = unique_grades[1]
    
    # 2. Get names of students with that grade
    qualifying_names = [name for name, score in records if score == second_lowest_grade]
    
    # 3. Print sorted alphabetically
    for name in sorted(qualifying_names):
        print(name)

# Demo
records = [
    ['Harry', 37.21],
    ['Berry', 37.21],
    ['Tina', 37.2],
    ['Akriti', 41.0],
    ['Harsh', 39.0]
]
second_lowest_students(records)
```
