### Key Concepts & Reference Notes
- **Data Types & Casting:** `int()`, `float()`, and `str()` explicitly convert types.
- **Precision Formatting:** `f'{val:.1f}'` formats floats to exactly one decimal place.
- **Operator Behavior:** `+` performs mathematical addition on numbers, but string concatenation on strings.
