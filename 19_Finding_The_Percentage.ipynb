# Finding The Percentage

**Difficulty:** Easy | **Topic:** hashmap, dict, data types

- [HackerRank Challenge](https://www.hackerrank.com/challenges/finding-the-percentage)

---

### Key Concepts & Reference Notes
- **Dictionary Mapping:** Name as key, list of numeric scores as value.
- **Average Calculation:** `sum(scores) / len(scores)`.
- **Strict Two-Decimal Formatting:** `f'{avg:.2f}'` ensures trailing zeros like `.00` are never omitted.

---

### Solution & Demonstration


```python
def print_average(student_marks, query_name):
    # Get marks list for queried student
    scores = student_marks[query_name]
    avg = sum(scores) / len(scores)
    # Print formatted strictly to 2 decimal places
    print(f'{avg:.2f}')

# Demo
marks = {
    'Krishna': [67.0, 68.0, 69.0],
    'Arjun': [70.0, 98.0, 63.0],
    'Malika': [52.0, 56.0, 60.0]
}
print_average(marks, 'Malika')
```
