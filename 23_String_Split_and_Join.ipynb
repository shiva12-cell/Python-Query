### Key Concepts & Reference Notes
- **`line.split(' ')`:** Splits the string by spaces into a list of words.
- **`'-'.join(...)`:** Joins an iterable of strings using a hyphen delimiter.
- **Composability:** Can be written cleanly in a single expression: `'-'.join(line.split(' '))`.

---

### Solution & Demonstration


```python
def split_and_join(line):
    # Split by spaces and rejoin with hyphens
    return '-'.join(line.split(' '))

# Demo
sample_line = 'this is a string'
print(split_and_join(sample_line))
```
