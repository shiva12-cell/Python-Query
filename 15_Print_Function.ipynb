### Key Concepts & Reference Notes
- **Unpacking with `*`:** Unpacks elements directly into `print()` arguments.
- **Custom Separator (`sep`):** `sep=''` joins unpacked arguments with no space between them.
- **Efficiency:** Avoids building intermediate string buffers in memory.

---

### Solution & Demonstration


```python
def print_consecutive(n):
    # Unpack range 1..n with empty separator
    print(*range(1, n + 1), sep='')

# Demo for n = 5 (outputs 12345)
print_consecutive(5)
```
