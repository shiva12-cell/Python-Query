### Key Concepts & Reference Notes
- **Percentage Calculation:** Tip and tax are percentages of base cost: `base * (percent / 100)`.
- **Rounding:** `round(total)` rounds floats to the nearest whole integer.
- **Function Encapsulation:** Putting the calculation in a clean helper function makes testing straightforward.

---

### Solution & Demonstration


```python
def solve(meal_cost, tip_percent, tax_percent):
    # Calculate tip and tax amounts
    tip = meal_cost * (tip_percent / 100)
    tax = meal_cost * (tax_percent / 100)
    
    # Total cost rounded to the nearest integer
    total_cost = round(meal_cost + tip + tax)
    print(total_cost)

# Sample test case
solve(meal_cost=12.00, tip_percent=20, tax_percent=8)
```
