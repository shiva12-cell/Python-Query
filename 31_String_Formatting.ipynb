### Key Concepts & Reference Notes
- **Number Base Conversions:** `oct(n)`, `hex(n)`, `bin(n)` return string representations with base prefixes.
- **Stripping Base Prefixes:** Slice `[2:]` removes `0o`, `0x`, `0b`.
- **Right Justification:** `.rjust(width)` right-aligns each number to match the binary width of the maximum number.

---

### Solution & Demonstration


```python
def print_formatted(number):
    # Max width is determined by the length of the binary string of 'number'
    width = len(bin(number)[2:])
    
    for i in range(1, number + 1):
        dec = str(i).rjust(width)
        octal = oct(i)[2:].rjust(width)
        hexa = hex(i)[2:].upper().rjust(width)
        binary = bin(i)[2:].rjust(width)
        print(f'{dec} {octal} {hexa} {binary}')

# Demo with number = 17
print_formatted(17)
```
