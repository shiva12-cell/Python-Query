# Designer Door Mat

**Difficulty:** Easy | **Topic:** str, format

- [HackerRank Challenge](https://www.hackerrank.com/challenges/designer-door-mat/problem)

---

### Key Concepts & Reference Notes
- **Step Ranges:** `range(1, n, 2)` produces odd multipliers `1, 3, 5...`.
- **Centering:** `pattern.center(m, '-')` centers the pattern with `-` padding on both sides.
- **Vertical Mirroring:** Use `range(n - 2, 0, -2)` for the bottom half to create a mirror of the top half.

---

### Solution & Demonstration


```python
def print_door_mat(n, m):
    # Top half: increasing odd repetitions of pattern '.|.'
    for i in range(1, n, 2):
        print(('.|.' * i).center(m, '-'))
    
    # Welcome center line
    print('WELCOME'.center(m, '-'))
    
    # Bottom half: decreasing odd repetitions
    for i in range(n - 2, 0, -2):
        print(('.|.' * i).center(m, '-'))

# Demo with N = 9, M = 27
print_door_mat(9, 27)
```
