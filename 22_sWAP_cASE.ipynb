### Key Concepts & Reference Notes
- **`str.swapcase()`:** Inverts the case of each alphabetic character in the string.
- **Non-Alpha Handling:** Non-letter characters (spaces, punctuation, digits) remain unchanged.
- **Immutability:** Strings in Python are immutable; `.swapcase()` returns a newly constructed string.
