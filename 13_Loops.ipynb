### Key Concepts & Reference Notes
- **`range(n)` Iteration:** Yields indices from 0 up to `n - 1`.
- **Exponentiation (`**`):** `i ** 2` squares `i` cleanly.
- **Loop Performance:** Standard `for` loop is optimized and readable.

---

### Solution & Demonstration


```python
def print_squares(n):
    # For all non-negative integers i < n, print i^2
    for i in range(n):
        print(i ** 2)

# Demo for n = 5
print_squares(5)
```
