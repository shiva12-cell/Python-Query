### Key Concepts & Reference Notes
- **The `__init__` Constructor:** Runs when a new instance is created to set initial attributes.
- **Instance State (`self`):** Variables prefixed with `self.` belong to the individual object instance.
- **Methods:** Functions defined inside a class that operate on the instance state.

---

### Solution & Demonstration


```python
class Person:
    def __init__(self, initialAge):
        # Age validation: negative ages are reset to 0
        if initialAge < 0:
            print('Age is not valid, setting age to 0.')
            self.age = 0
        else:
            self.age = initialAge

    def amIOld(self):
        # Output age classification
        if self.age < 13:
            print('You are young.')
        elif 13 <= self.age < 18:
            print('You are a teenager.')
        else:
            print('You are old.')

    def yearPasses(self):
        # Age increases by 1 each year
        self.age += 1

# Demo test
p = Person(4)
p.amIOld()
for _ in range(10):
    p.yearPasses()
p.amIOld()
```
