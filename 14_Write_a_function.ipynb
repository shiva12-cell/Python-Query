### Key Concepts & Reference Notes
- **Function Definition:** `def is_leap(year):` defines reusable logic returning a boolean.
- **Leap Year Rules:** Divisible by 4, unless divisible by 100, unless also divisible by 400.
- **Boolean Operators:** Using `and` and `or` with proper precedence allows a clean one-line check.

---

### Solution & Demonstration


```python
def is_leap(year):
    # Year divisible by 4 and not 100, or divisible by 400
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# Demo test cases
years = [1990, 2000, 2100, 2024]
for y in years:
    print(f'{y}: is_leap = {is_leap(y)}')
```
