### Key Concepts & Reference Notes
- **Reversing via Slice:** `arr[::-1]` creates a reversed copy without modifying original list.
- **Argument Unpacking (`*`):** `print(*arr)` prints space-separated elements without manual string conversion.
- **Parsing Input List:** `list(map(int, input().split()))` reads integers into a list in one line.

---

### Solution & Demonstration


```python
def print_reversed(arr):
    # Reverse array using slice and unpack into print
    print(*arr[::-1])

# Demo
sample_arr = [1, 4, 3, 2]
print('Original:', sample_arr)
print('Reversed: ', end='')
print_reversed(sample_arr)
```
