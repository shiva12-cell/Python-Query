### Key Concepts & Reference Notes
- **`textwrap.fill(text, width)`:** Formats a single string into lines of maximum width separated by `\n`.
- **`textwrap.wrap(text, width)`:** Returns a list of lines instead of a single string.
- **Built-in Standard Library:** Part of Python's standard library, no external packages needed.

---

### Solution & Demonstration


```python
import textwrap

def wrap(string, max_width):
    # textwrap.fill splits text into max_width chunks joined by newlines
    return textwrap.fill(string, max_width)

# Demo
sample_str = 'ABCDEFGHIJKLIMNOQRSTUVWXYZ'
print(wrap(sample_str, 4))
```
