### Key Concepts & Reference Notes
- **`range(start, stop)`:** `range(1, 11)` produces numbers from 1 to 10 inclusive.
- **Iteration:** `for i in range(...)` iterates over each number sequentially.
- **f-Strings:** Cleaner than concatenation when building formatted arithmetic strings.

---

### Solution & Demonstration


```python
def print_multiples(n):
    # Print the first 10 multiples: n x i = result
    for i in range(1, 11):
        print(f'{n} x {i} = {n * i}')

# Demo with n = 2
print_multiples(2)
```
