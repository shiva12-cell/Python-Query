### Key Concepts & Reference Notes
- **Step Slicing:** `s[start:stop:step]` allows skipping characters.
- **Even Indices:** `s[::2]` grabs characters at indices 0, 2, 4...
- **Odd Indices:** `s[1::2]` grabs characters at indices 1, 3, 5...

---

### Solution & Demonstration


```python
def separate_even_odd_chars(s):
    # Using step 2 slicing to separate indices
    even_chars = s[::2]
    odd_chars = s[1::2]
    print(f'{even_chars} {odd_chars}')

# Demo test cases
for word in ['Hacker', 'Rank']:
    separate_even_odd_chars(word)
```
