# Text Alignment

**Difficulty:** Easy | **Topic:** str, format

- [HackerRank Challenge](https://www.hackerrank.com/challenges/text-alignment/problem)

---

### Key Concepts & Reference Notes
- **Formatting Methods:** `.ljust(width)` left aligns, `.rjust(width)` right aligns, `.center(width)` centers.
- **String Multiplication:** Repeating characters with `*` creates geometric lines.
- **HackerRank Logo Structure:** Constructed from top cone, top pillars, center belt, bottom pillars, and bottom cone.

---

### Solution & Demonstration


```python
def draw_logo(thickness):
    c = 'H'
    
    # Top Cone
    for i in range(thickness):
        print((c*i).rjust(thickness-1) + c + (c*i).ljust(thickness-1))
    
    # Top Pillars
    for i in range(thickness+1):
        print((c*thickness).center(thickness*2) + (c*thickness).center(thickness*6))
    
    # Middle Belt
    for i in range((thickness+1)//2):
        print((c*thickness*5).center(thickness*6))
    
    # Bottom Pillars
    for i in range(thickness+1):
        print((c*thickness).center(thickness*2) + (c*thickness).center(thickness*6))
    
    # Bottom Cone
    for i in range(thickness):
        print(((c*(thickness-i-1)).rjust(thickness) + c + (c*(thickness-i-1)).ljust(thickness)).rjust(thickness*6))

# Demo with thickness = 5
draw_logo(5)
```
