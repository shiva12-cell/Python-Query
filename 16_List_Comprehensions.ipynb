### Key Concepts & Reference Notes
- **Multi-Variable Comprehensions:** Nest multiple `for` clauses inside `[...]` from outer to inner.
- **Conditional Filter (`if`):** Place condition at the end to filter elements.
- **Clean Syntax:** Replaces 3 nested `for` loops and `.append()` calls with a single readable expression.

---

### Solution & Demonstration


```python
def generate_coordinates(x, y, z, n):
    # Cartesian coordinates [i, j, k] where sum != n
    return [
        [i, j, k]
        for i in range(x + 1)
        for j in range(y + 1)
        for k in range(z + 1)
        if (i + j + k) != n
    ]

# Demo
print(generate_coordinates(1, 1, 1, 2))
```
